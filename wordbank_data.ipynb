```python
from pandas_datareader import wb
import pandas_datareader as pdr
import pandas as pd
import numpy as np
```

```python
# Download data from World Bank
# Indicators: GDP (current US$), CPI (annual %), Population growth (annual %)
# Countries: Vietnam (VNM), Afghanistan (AFG), Albania (ALB), Anguilla (AIA), Angola (AGO) (Note: Original image had AFE, which is not a valid WB country code, I've used AFG for Afghanistan as a common alternative. If you intended a different country, please provide the correct code.)
# Start year: 1996, End year: 2022
GDP_VN = wb.download(indicator=['NY.GDP.MKTP.CD', 'FP.CPI.TOTL.ZG', 'SP.POP.GROW'],
                     country=['VNM', 'AFG', 'ALB', 'AIA', 'AGO'], # Adjusted based on common WB codes
                     start=1996,
                     end=2025)
GDP_VN
```

```python
GDP_VN.colums=['GDP_VN', 'CPI_VN', 'POP_GROW_VN']
GDP_VN
```

```python
GDP_VN = GDP_VN.reset_index()
GDP_VN.head()
```

```python
# Save the DataFrame to an Excel file with the specified filename
GDP_VN.to_excel('GDP1.xlsx')
```
